```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import os
```

```python
input = {
        "12stem": "alex/oxview_12nt/main",
        "9stem": "alex/oxview_9nt/main",
        "6stem": "alex/oxview_6nt/main",
        "3stem": "alex/oxview_3nt/main"
}

def load_hb_count(path: str) -> pd.DataFrame:
    return pd.read_csv(path, sep=r'\s+', header=None, names=["t", "hb_count"], index_col="t")

dfs = {}
for stem, path in input.items():
    dfs[stem] = load_hb_count(os.path.join(path, "hb_count.dat"))
    print(dfs[stem].head())
```

```python
plt.figure(figsize=(12, 4))

for stem in input:
    plt.plot(dfs[stem], label = stem)
plt.xlabel("Step")
plt.ylabel("Hydrogen bonds formed")
plt.legend()
```
