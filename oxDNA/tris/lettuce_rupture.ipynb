```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os

# Configure matplotlib to output both SVG (vector) and PNG (raster) formats
%config InlineBackend.figure_formats = ['svg']
```

```python
INPUT_PATHS = ["lettuce_12nt/rupture_pushing_full_rod\\lettuce_hbonds.dat",
               "lettuce_6nt/rupture_pushing_full_rod\\lettuce_hbonds.dat"]

hbonds_dfs = {}
for path in INPUT_PATHS:
    if "12nt" in path:
        hbonds_dfs["12nt"] = pd.read_csv(path, delim_whitespace=True, header=None, names=["time", "h_bonds"])
    elif "6nt" in path:
        hbonds_dfs["6nt"] = pd.read_csv(path, delim_whitespace=True, header=None, names=["time", "h_bonds"])
```

```python
#### GLOBAL PLOT CONFIGURATION
plt.rcParams.update({
    "axes.linewidth": 1.5,
    "xtick.major.width": 1.5,
    "ytick.major.width": 1.5,
    "xtick.major.size": 8,
    "ytick.major.size": 8,
    "font.size": 12,
    "axes.labelsize": 12,
    "legend.fontsize": 12,
})
```

```python
#### PLOT OF HBONDS IN LETTUCE STEM OVER TIME
fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(hbonds_dfs["12nt"]["time"], hbonds_dfs["12nt"]["h_bonds"], 
        label="12nt Lettuce", color='blue', linewidth=0.75)
ax.plot(hbonds_dfs["6nt"]["time"], hbonds_dfs["6nt"]["h_bonds"], 
        label="6nt Lettuce", color='green', linewidth=0.75)
ax.set_xlabel("Time (epochs)")
ax.set_ylabel("Number of Hydrogen Bonds")
ax.set_ylim(0, 12.5)
```
