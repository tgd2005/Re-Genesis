```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os

# Configure matplotlib to output both SVG (vector) and PNG (raster) formats
%config InlineBackend.figure_formats = ['svg']
```

```python
INPUT_PATH = "lettuce_12nt/rupture_pushing_full_rod\\lettuce_hbonds.dat"

lettuce12_hb_df = pd.read_csv(INPUT_PATH, delim_whitespace=True, header=None, names=["time", "h_bonds"])
```

```python
fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(lettuce12_hb_df["time"], lettuce12_hb_df["h_bonds"], 
        label="12nt Lettuce", color='blue', linewidth=0.75)
ax.set_xlabel("Time (epochs)", fontsize=12)
ax.set_ylabel("Number of Hydrogen Bonds", fontsize=12)
ax.set_ylim(0, 12.5)
```
